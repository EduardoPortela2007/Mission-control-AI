```python
import ollama
import random
```

```python
modo_teste = False #para usar o modo_teste é só trocar o False por True
if modo_teste:

    temperatura = 115
    energia = 8
    oxigenio = 45
    comunicacao = "Sem Sinal"
    modulo = "Falha"

else:

    temperatura = random.randint(20,120)

    energia = random.randint(5,100)

    oxigenio = random.randint(40,100)

    comunicacao = random.choice([
        "Estável",
        "Instável",
        "Sem Sinal"
    ])

    modulo = random.choice([
        "Operacional",
        "Manutenção",
        "Falha"
    ])
```

```python
alertas = []

if temperatura > 90:
    alertas.append("Temperatura crítica")

if energia < 20:
    alertas.append("Energia muito baixa")

if oxigenio < 60:
    alertas.append("Oxigênio abaixo do ideal")

if comunicacao != "Estável":
    alertas.append("Problema de comunicação")

if modulo == "Falha":
    alertas.append("Falha detectada no módulo")

if len(alertas) == 0:
    alertas.append("Nenhum alerta encontrado")
```

```python
acoes = []

if temperatura > 90:
    acoes.append("Ativar sistema de resfriamento")

if energia < 20:
    acoes.append("Ativar modo economia")

if oxigenio < 60:
    acoes.append("Liberar reserva de oxigênio")

if comunicacao == "Sem Sinal":
    acoes.append("Iniciar comunicação emergencial")

if modulo == "Falha":
    acoes.append("Enviar equipe de manutenção")

if len(acoes) == 0:
    acoes.append("Nenhuma ação necessária")
```

```python
pontos_risco = 0

if temperatura > 90:
    pontos_risco += 1

if energia < 20:
    pontos_risco += 1

if oxigenio < 60:
    pontos_risco += 1

if comunicacao != "Estável":
    pontos_risco += 1

if modulo == "Falha":
    pontos_risco += 1

    if pontos_risco <= 1:
      nivel = "NORMAL"

elif pontos_risco <= 3:
    nivel = "ATENÇÃO"

else:
    nivel = "CRÍTICO"
```

```python
resposta = ollama.chat(
    model="llama3.2:1b",
    messages=[
        {
            "role":"system",
            "content":
            """
            Você é um especialista em controle de missão espacial.
            Analise os dados recebidos.
            Explique os problemas encontrados.
            Sugira melhorias.
            """
        },
        {
            "role":"user",
            "content":
            f"""
            Temperatura: {temperatura}
            Energia: {energia}
            Oxigênio: {oxigenio}
            Comunicação: {comunicacao}
            Módulo: {modulo}
            """
        }
    ]
)

analise_ia = resposta["message"]["content"]
```

```python
def mostrar_relatorio():

    print("="*60)
    print("MISSION CONTROL AI")
    print("="*60)

    print(f"Temperatura : {temperatura} °C")
    print(f"Energia     : {energia}%")
    print(f"Oxigênio    : {oxigenio}%")
    print(f"Comunicação : {comunicacao}")
    print(f"Módulo      : {modulo}")

    print("\nNível da Missão:")
    print(nivel)

    print("\nAlertas:")

    for alerta in alertas:
        print("-", alerta)

    print("\nAções:")

    for acao in acoes:
        print("-", acao)

    print("\nAnálise da IA:")
    print(analise_ia)

    print("="*60)
```

```python
mostrar_relatorio()
```
